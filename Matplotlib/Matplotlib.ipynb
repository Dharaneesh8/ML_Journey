```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 250, 300, 280, 350]

plt.plot(months, sales)
plt.xticks(rotation=90)

plt.annotate(
    "Lowest sales",
    xy=("Jan", 200),
    xytext=("Jan", 340),
    arrowprops=dict(arrowstyle="->")
)

plt.title("Monthly Sales Data with Annotation")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("summa.jpg")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 250, 180, 300, 350]
profit = [50, 70, 40, 90, 110]

# First chart
plt.subplot(2, 1, 1)

plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.ylabel("Sales")


# Second chart
plt.subplot(2, 1, 2)

plt.bar(months, profit)
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [200, 250, 180, 300, 350]
profit = [50, 70, 40, 90, 110]
expenses = [150, 180, 140, 210, 240]

fig, ax = plt.subplots(2, 2)

ax[0, 0].plot(months, sales)
ax[0, 0].set_title("Sales")

ax[0, 1].bar(months, profit)
ax[0, 1].set_title("Profit")

ax[1, 0].plot(months, expenses)
ax[1, 0].set_title("Expenses")

ax[1, 1].scatter(sales, profit)
ax[1, 1].set_title("Sales vs Profit")

plt.tight_layout()
plt.show()
```
