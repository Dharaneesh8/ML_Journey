```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Arun", "Bala", "Kiran", "Divya", "Meena", "Rahul", "Priya", "Vijay"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "IT"],
    "Salary": [50000, 65000, 72000, 80000, 58000, 90000, 55000, 85000],
    "Experience": [1, 5, 3, 12, 4, 10, 2, 8]
})
```

```python
dept_sal=df.groupby("Department")["Salary"].mean()
print(dept_sal)
```

```python
dept_sal.plot.bar()
plt.title("Sample")
plt.xlabel("Department")
plt.ylabel("Salary")
```

```python
df["Department"].value_counts().plot.bar()
plt.title("Count")
plt.xlabel("Department")
plt.ylabel("count of employees")
```

```python
dept=df.groupby("Department").agg(
    Avg_Salary=("Salary",'mean'),
    Avg_experience=("Experience",'mean'),
    Employee_count=("Name",'count')
)
print(dept)
```

```python
dept["Avg_Salary"].plot.bar()

plt.title("Avg Salary")
plt.xlabel("Department")
plt.xticks(rotation=45)
plt.ylabel("Avg salary")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

dept["Avg_Salary"].plot(kind="bar", ax=ax[0])
ax[0].set_title("Average Salary")
ax[0].set_xlabel("Department")
ax[0].set_ylabel("Salary")

dept["Employee_count"].plot(kind="bar", ax=ax[1])
ax[1].set_title("Employee Count")
ax[1].set_xlabel("Department")
ax[1].set_ylabel("Employees")

plt.tight_layout()
plt.show()
```
