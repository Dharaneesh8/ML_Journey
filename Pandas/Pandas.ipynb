```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [" Arun ", "Bala", " Kiran", "Divya", "Bala"],
    "Department": ["it-dept", "HR", "IT-DEPT", "hr ", "HR"],
    "Age": [24, 31, 27, 42, 31],
    "Salary": [50000, 65000, 72000, 80000, 65000],
    "Experience": [1, 5, 3, 12, 5],
    "Performance": [4, 3, 5, 4, 3]
})
```

```python
df.head()
df.shape
df.info()
df.isnull().sum()
df.duplicated().sum()
```

```python
df["Name"]=df["Name"].str.strip()
```

```python

df["Department"]=(df["Department"]
                  .str.upper()
                  .str.replace("-DEPT","" ,regex=False)

                  .str.strip()
)
```

```python
df = df.drop_duplicates()
```

```python
df["New Salary"]=df["Salary"]*1.10
```

```python
df["Experience_Level"]=np.where(
    df["Experience"]>=5,
    "Experienced",
    "Fresher"
)
```

```python
dept_salary = df.groupby("Department")["Salary"].mean()
print(dept_salary)
```

```python
dept_count = df["Department"].value_counts()
print(dept_count)
```

```python
highest_paid = df.loc[df["Salary"].idxmax()]
```

```python
df.groupby("Experience_Level")["Salary"].mean()
```

```python
df["Dept_Avg_Salary"] = (
    df.groupby("Department")["Salary"]
      .transform("mean")
)
print(df)
```

```python
above_average = df[
    df["Salary"] > df["Dept_Avg_Salary"]
]
```

```python
df.to_csv("clean_employee_data.csv", index=False)
```
