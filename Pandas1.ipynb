```python
import pandas as pd

# 1. Create a sample DataFrame
data = {
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Department": ["Sales", "Engineering", "Sales", "Engineering", "Engineering"],
    "Salary": [45000, 80000, 40000, 95000, 65000],
    "Age": [20,30,50,22,33]
}
df = pd.DataFrame(data)

result=df.groupby("Department").filter(
    lambda x:x["Salary"].mean()>50000
)
print(result)
```

```python
df["Age_Groups"]=pd.cut(
    df["Age"],
    bins=[0,25,45,70],
    labels=["Young","Adult","Senior Citizen"]
)
print(df)
```

```python
df["Package"]=pd.qcut(
    df["Salary"],
    q=2,
    labels=["low","high"]
)
print(df)
```

```python
import pandas as pd

df=pd.DataFrame({
    "Dept":["IT","HR","HR","IT"],
    "Salary":[45000,50000,20000,40000],
    "Year":[2024,2025,2024,2025]
})

res=df.groupby(["Dept","Year"])["Salary"].mean()
print(res)
```

```python
poda=res.unstack()
print(poda)

vada=poda.stack()
print(vada)
```

```python
poda=res.unstack(level="Dept")
print(poda)
```

```python
df1 = pd.DataFrame({
    "Employee_ID": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Department": ["HR", "Engineering", "Sales", "HR"]
})

df2 = pd.DataFrame({
    "Employee_ID": [3, 4, 5, 6],
    "Salary": [70000, 95000, 60000, 85000],
    "Project": ["Alpha", "Beta", "Gamma", "Delta"]
})

result = pd.merge(
    df1,
    df2,
    on="Employee_ID",
    how="outer",
    indicator=True
)
print(result)
```

```python
result[result["_merge"] == "left_only"]
```

```python
import pandas as pd

# Define df to ensure it's available for the Excel operations
data = {
    "Employee": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Department": ["Sales", "Engineering", "Sales", "Engineering", "Engineering"],
    "Salary": [45000, 80000, 40000, 95000, 65000],
    "Age": [20,30,50,22,33]
}
df = pd.DataFrame(data)

with pd.ExcelWriter("employee_report.xlsx") as writer:
    df.to_excel(writer, sheet_name="Employees", index=False)
    df.groupby("Department")["Salary"].mean().to_excel(
        writer,
        sheet_name="Department_Salary"
    )
```
