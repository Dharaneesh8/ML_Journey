```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder

df=pd.DataFrame({
    "Age":[25,30,35,40],
    "Experience":[1,5,10,15],
    "Department":["HR","IT","Finance","HR"]
})

preprocessor=ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),["Age","Experience"]),
        ("cat",OneHotEncoder(),["Department"]) #one hot encoding is used to convert categorical data into numerical data
    ]
)

x_transformed=preprocessor.fit_transform(df)
print(x_transformed)
```

```python
from sklearn.preprocessing import OneHotEncoder

df=pd.DataFrame({
    "City":["Chennai","Bangalore","Chennai","Delhi"]
})

encoder=OneHotEncoder(sparse_output=False) #sparse=False is used to return a dense array instead of a sparse matrix
x_encoded=encoder.fit_transform(df[["City"]])
print(encoder.get_feature_names_out())
print(x_encoded)
```

ORDINALENCODER is used when we need to label the categories in ordered

```python
from sklearn.preprocessing import OrdinalEncoder

df=pd.DataFrame({
    "Satisfaction":["High","Medium","Low","Medium"]
})

encoded=OrdinalEncoder(
    categories=[["Low","Medium","High"]], #specifies the order of the categories
)

scaled=encoded.fit_transform(df[["Satisfaction"]])
print(scaled)
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder

df=pd.DataFrame({
    "Age":[25,30,35,40],
    "City":["Chennai","Bangalore","Chennai","Delhi"],
    "Satisfaction":["High","Medium","Low","Medium"]
})

encoder=ColumnTransformer(
    transformers=[
        ("num",StandardScaler(),["Age"]),
        ("nominal",OneHotEncoder(),["City"]),
        ("ordinal",OrdinalEncoder(categories=[["Low","Medium","High"]]),["Satisfaction"])
    ]
)

scaled=encoder.fit_transform(df)
print(scaled)
```
