```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X=np.array([[1],[2],[3],[4],[5],[6]])
y=np.array([0,0,0,1,1,1])

model=LogisticRegression()
model.fit(X,y)

print(model.predict([[3.5]]))  #this predicts whether pass or fail using 0 and 1
print(model.predict_proba([[4.5]])) #this give the probablity of getting 0 and 1 it returns two values first value is the probability of getting 0 and second value is the probability of getting 1
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

y_true=np.array([0,0,0,1,1,1])
y_pred=np.array([1,0,1,1,1,1])

print("Accuracy:",accuracy_score(y_true,y_pred)) #Accuracy	How many total predictions were correct?
print("Precision:",precision_score(y_true,y_pred)) #Precision	How many predicted positives were actually positive?
print("Recall:",recall_score(y_true,y_pred)) #Recall	How many actual positives did we identify?
print("F1-Score:",f1_score(y_true,y_pred)) #F1-Score	How well the model balances precision and recall
```
