```python
from sklearn.linear_model import LinearRegression
import pandas as pd

df = pd.DataFrame({
    "StudyHours": [2, 4, 6, 8],
    "Attendance": [60, 75, 80, 95],
    "Marks": [45, 65, 78, 92]
})

X = df[["StudyHours", "Attendance"]]
y = df["Marks"]

model = LinearRegression()
model.fit(X, y)

prediction = model.predict(
    pd.DataFrame({
        "StudyHours": [9],
        "Attendance": [80]
    })
)

print(prediction)
print(model.coef_)
print(model.intercept_)
```
