Simple Imputer is used to handle the missing values category wise like mean,median,most_frequent,

```python
import numpy as np
from sklearn.impute import SimpleImputer

X=np.array([[20],[30],[np.nan],[40]])

imputer=SimpleImputer(strategy='mean') #calculates mean of the column and replaces the missing values with mean
x_filled=imputer.fit_transform(X)

print(x_filled)
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer

df=pd.DataFrame({
    "Department":["IT","HR",np.nan,"HR","IT"],
})

imputer=SimpleImputer(strategy="most_frequent")
#filled=imputer.fit_transform(df)     (this gives in 2D array)
#print(filled)

df["Department"]=imputer.fit_transform(df[["Department"]]).ravel() #here ravel is used to convert to 1D array
print(df)
```
